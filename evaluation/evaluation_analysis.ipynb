```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_palette('Pastel2')
```

LLM-as-a-judge results

```python
data = pd.read_csv('/Users/druzzzsk/projects/biocad/evaluation/llm-as-a-judge/judge_results.csv')
data.head()
```

```python
criteria = ['factual_consistency', 'coverage', 'citation', 'paraphrase', 'relevance']

plt.figure(figsize=(10, 6))
sns.boxplot(data=data[criteria])
plt.title('Распределение оценок по критериям')
plt.ylabel('Оценка')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
data[criteria].mean().plot(kind='bar')
plt.title('Средние оценки по критериям')
plt.ylabel('Средняя оценка')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylim(0, 5)
plt.show()
```
